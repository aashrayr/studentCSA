1. Why does changing b not affect a, but changing array2 affects array1?
    Changing b doesn’t change a because a and b each hold their own copy of the number. But array1 and array2 both point to the same list in memory, so when you change one, the other changes too, they share the same data.

2. Describe what's on the stack vs. the heap for this code.
    The stack holds simple things like numbers and the names of variables (a, b, array1, array2). The heap holds bigger things like the actual array {99, 2, 3}. The stack just keeps the address (a pointer) that tells the program where the array is in the heap.

## Homework Hack #2 — Heap vs Stack Storage Demo

**Goal** - Understand where data is stored in memory — stack vs heap.
<br>
Instructions:
- Create a Book class with one variable: String title.
- In main(), create:
- A primitive variable (e.g. int pages = 300;)
- A Book object (e.g. Book b1 = new Book("Java Basics");)
- Copy both into new variables (int pagesCopy = pages;, Book b2 = b1;)
<br>
Change the original values and print everything — watch what changes.

```java
// Homework Hack #2: Heap vs Stack Storage Demo

class Book {
    private String title;

    public Book(String title) {
        this.title = title;
    }

    public void setTitle(String title) {
        this.title = title;
    }

    public String toString() {
        return "Book Title: " + title;
    }
}

class HeapVsStack {
    public static void run() {
        int pages = 300;
        int pagesCopy = pages;

        Book b1 = new Book("Java Basics");
        Book b2 = b1;

        System.out.println("Before changes:");
        System.out.println("pages = " + pages + ", pagesCopy = " + pagesCopy);
        System.out.println("b1 = " + b1 + ", b2 = " + b2);

        pages = 500;
        b1.setTitle("Advanced Java");

        System.out.println("\nAfter changes:");
        System.out.println("pages = " + pages + ", pagesCopy = " + pagesCopy);
        System.out.println("b1 = " + b1 + ", b2 = " + b2);
    }
}

HeapVsStack.run();

```
