# Homework: Ultimate Battle

## Scenario

You are programming a game where 2 objects of your choosing (e.g., Robots, Dinosaurs, People) battle. Each object has health and power.

---

## Instructions

1. **Create a class** representing your object with:

   **Instance variables:**  
   - `String name`  
   - `int power`  
   - `int health`  
   - `MAKE YOUR OWN`

   **Static variables:**
   - `double fightDuration`

   **Instance methods:**  
   - `void attack()`
   - `void printStatus()`
   - `MAKE YOUR OWN`

   **Class methods:**  
   - `static int strongerFighter()`
   - `static void beginBattle()`
   - `MAKE YOUR OWN`

2. **In `main`:**  
   - Create two objects.  
   - Use instance methods to attack and print status.  
   - Use static methods to compare, print a fact, and start the battle.

```java
// Code Here
public class UltimateBattle {
    static class Warrior {
        String name;
        int power;
        int health;
        String weapon;
        static double fightDuration = 0.0;

        Warrior(String name, int power, int health, String weapon) {
            this.name = name;
            this.power = power;
            this.health = health;
            this.weapon = weapon;
        }

        void attack(Warrior opponent) {
            System.out.println(name + " attacks " + opponent.name + " with " + weapon + "!");
            int damage = (int) (Math.random() * power);
            opponent.health -= damage;
            if (opponent.health < 0) opponent.health = 0;
            System.out.println(name + " dealt " + damage + " damage!");
        }

        void printStatus() {
            System.out.println(name + " | Health: " + health + " | Power: " + power + " | Weapon: " + weapon);
        }

        void heal(int amount) {
            health += amount;
            System.out.println(name + " heals for " + amount + " points!");
        }

        static Warrior strongerFighter(Warrior a, Warrior b) {
            return (a.power > b.power) ? a : b;
        }

        static void beginBattle(Warrior a, Warrior b) {
            System.out.println("\nBattle begins between " + a.name + " and " + b.name + "\n");
            fightDuration = Math.random() * 5 + 1;
            while (a.health > 0 && b.health > 0) {
                a.attack(b);
                if (b.health <= 0) break;
                b.attack(a);
                System.out.println();
            }
            if (a.health <= 0 && b.health <= 0)
                System.out.println("It's a draw!");
            else if (a.health <= 0)
                System.out.println(b.name + " wins the battle!");
            else
                System.out.println(a.name + " wins the battle!");
            System.out.printf("Battle lasted %.2f minutes.%n", fightDuration);
        }

        static void printBattleFact() {
            System.out.println("Fun Fact: The stronger fighter usually wins, but luck matters too!");
        }
    }

    public static void main(String[] args) {
        Warrior w1 = new Warrior("Ares", 30, 120, "Sword");
        Warrior w2 = new Warrior("Zeus", 25, 140, "Thunderbolt");
        w1.printStatus();
        w2.printStatus();
        Warrior.printBattleFact();
        Warrior stronger = Warrior.strongerFighter(w1, w2);
        System.out.println("Stronger fighter: " + stronger.name);
        Warrior.beginBattle(w1, w2);
        System.out.println("\nFinal Status:");
        w1.printStatus();
        w2.printStatus();
    }
}

UltimateBattle.main(null);
```
